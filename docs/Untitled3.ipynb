# schema are functions

schema are the hidden structure and lattices that describe objects.
our concern here are explicit schema that delineate the namespace for a problem.

a context-free object exists without schema, the keys and values are as arbitary as any other text. including schema and describing the type space adds context to the keys, and the values.


a schema delineates the namespace for an object it estabilishes the vocabulary, keys and values, words and definitions.

```python
from nbref import meta
```

```python
thing = meta.Dict(foo="a really important thing").modify(
    title="a representitive object",
    properties=dict(
        foo=dict(
            description="a misleadingly un-important thing",
            type="string"
        )
    )
).validate()
```

```python
thing
```

```python
thing.schema(
    foo=1
).errors()
```

```python
thing.display()
```
