# references

a specific feature of schema is explicit access to the web and other resources internal and external to the schema.

the document checks the ways that can schema can referenced.

there are context free and context dependent references.
referencing files and external documents are context free
meanwhile internal document references and pointers are context
dependent.

speicifcally for schema we can provide strong caching mechanisms

```python
from nbref import Schema, meta
testing = unittest.TestCase()
```

## context free references

### local path and file references

```python
assert Schema.Ref("file://./refs.ipynb").resolve()
```

### remote references

```python
assert Schema.Ref("https://api.github.com/users/tonyfast/gists?per_page=2").resolve()
```

## context dependent references

context dependent references require a root schema to  reference.

```python
Schema.Ref("#/$defs/foo")
```

```python
with testing.assertRaises(BaseException): 
    input, "requires a root to reference the pointer"
    Schema.Ref("#/$defs/foo").resolve()
```

## dynamic references

dynamic references dependent on dynamic anchors that allow for extending schema.

```python
assert Schema.Ref("#meta").set_root(meta)

with testing.assertRaises(BaseException):
    Schema.Ref("#meta").resolve()
```

```python
assert Schema.Ref("#meta").set_root(
    Schema("https://json-schema.org/draft/2020-12/meta/applicator")
)
```
