## similarities between schema and data, type and object

* both schema and data are objects
* the location, or cursor, is always known
* data and schema are constrained by schema

## differences betweens schema and data

* schema contains internal references, data does not
* in the moment of a problem statement, the schema has a constant shape, and the data does not
* schema lacks form, data has form. when the schema is being represented it sublimates to data.

`````python
%%
````js
function toggleClass(event) {
    // check for the input type, it might be select or input checkbox
    var el = event.target;
    if (el.tagName == "INPUT") {
        if (el.type == "checkbox") {
            (el.ariaControlsElements || []).forEach((target) => {
                target.classList.toggle(el.checked)
            })
        }
    } else if (el.tagName == "SELECT") {
        (el.ariaControlsElements || []).forEach((target) => {
            Array.from(el.options).forEach((option) => {
                target.classList.toggle(option.value, option.checked) 
            })
            
        })
    }
}
````
`````
