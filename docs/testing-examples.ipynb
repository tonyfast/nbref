## the not equivalence

the not equivalence describes the empty/null space. bad examples may be defined and the `not` can be added to describes inconsistent inputs.

```yaml
type: boolean
default: true
examples:
- "true"
- 1
```

```yaml
type: boolean
default: true
examples:
- "true"
- 1
not:
    type: [string, number]
```
